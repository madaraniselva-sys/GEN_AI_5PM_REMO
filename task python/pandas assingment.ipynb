```python
import pandas as pd
import numpy as np
data = {
    'EmpID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya', 'Rahul',
             'Meena', 'Sathish', 'Nila', 'Vijay', 'Kavya'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT',
                   'HR', 'Finance', 'IT', 'Sales', 'Sales'],
    'Age': [25, 29, 32, 27, 24, 35, 31, 28, 30, 26],
    'Salary': [35000, 42000, 58000, 39000, 30000,
               62000, 51000, 47000, 44000, 36000],
    'Experience': [2, 5, 8, 3, 1, 10, 7, 4, 6, 2],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Coimbatore',
             'Salem', 'Chennai', 'Madurai', 'Coimbatore',
             'Chennai', 'Salem']
}

df = pd.DataFrame(data)
```

1.Load and inspect the data

Create a Pandas DataFrame from the sample dataset. Display the first 5 rows, last 3 rows, the number of rows and columns, and the column data types.

```python
print("First 5 rows:")
print(df.head(5))

print("\nLast 3 rows:")
print(df.tail(3))

print("\nNumber of rows and columns:")
print(df.shape)

print("\nColumn data types:")
print(df.dtypes)
```

2.Filter employee records

 Display the details of employees who work in the IT department and have a salary greater than 40,000.

```python
result = df[(df["Department"] == "IT") &
            (df["Salary"] > 40000)]

print(result)
```

3.Sort the data

 Display all employees sorted by salary from highest to lowest. Show only Name, Department, and Salary.

```python
result = df.sort_values(
    by="Salary",
    ascending=False
)

print(result[["Name", "Department", "Salary"]])
```

4.Calculate summary statistics

 Find the mean, maximum, minimum, and median salary. Also find the average age of all employees.

```python
print("Mean Salary:", df['Salary'].mean())
print("Maximum Salary:", df['Salary'].max())
print("Minimum Salary:", df['Salary'].min())
print("Median Salary:", df['Salary'].median())
print("Average Age:", df['Age'].mean())
```

5.Group by department

 For each department, calculate the number of employees, average salary, and highest salary. Sort the result by average salary in descending order.

```python
result = df.groupby('Department').agg(
    EmployeeCount=('EmpID', 'count'),
    AverageSalary=('Salary', 'mean'),
    HighestSalary=('Salary', 'max')
).sort_values(by='AverageSalary', ascending=False)

print(result)
```

6.Create a new column

 Create a new column named AnnualSalary by multiplying Salary by 12. Display Name, Salary, and AnnualSalary.

```python
df['AnnualSalary'] = df['Salary'] * 12

print(df[['Name', 'Salary', 'AnnualSalary']])
```

7.Find missing and duplicate values

 Make a copy of the DataFrame, set one employee's City value to missing, and add a duplicate row.
 Use Pandas to identify missing values, fill the missing City with 'Unknown', and remove duplicate rows.

```python
df2 = df.copy()

df2.loc[0, 'City'] = np.nan

df2 = pd.concat([df2, df2.iloc[[0]]], ignore_index=True)

print("Missing values:")
print(df2.isnull().sum())

df2['City'] = df2['City'].fillna('Unknown')

df2 = df2.drop_duplicates()


print("\nAfter cleaning:")
print(df2[['Name', 'City']])
print("Rows:", len(df2))
```

8.Use conditional logic

 Create a column named SalaryLevel: 'High' for salary greater than or equal to 50,000, 'Medium' for
salary from 40,000 to 49,999, and 'Low' for salary below 40,000. Display Name, Salary, and SalaryLevel.

```python
def salary_level(salary):
    if salary >= 50000:
        return 'High'
    elif salary >= 40000:
        return 'Medium'
    else:
        return 'Low'

df['SalaryLevel'] = df['Salary'].apply(salary_level)

print(df[['Name', 'Salary', 'SalaryLevel']])
```

9.Work with a second DataFrame

 Create a marks DataFrame with columns EmpID and TrainingScore, containing scores for EmpID 101, 103, 105, 108, and 110. Merge it with the employee
DataFrame using EmpID. Display all employees and their scores, including employees without a score.

```python
marks = pd.DataFrame({
    'EmpID': [101, 103, 105, 108, 110],
    'TrainingScore': [85, 92, 70, 88, 78]
})

result = pd.merge(df, marks, on='EmpID', how='left')

print(result[['EmpID', 'Name', 'TrainingScore']])
```

10.Answer a business question

Find the department with the highest average salary. Then display the employees whose salary is above the overall average salary.
    Write 2–3 sentences explaining your findings.

```python
avg_salary = df['Salary'].mean()

dept_avg = df.groupby('Department')['Salary'].mean()

print("Department with highest average salary:")
print(dept_avg.idxmax())

print("\nHighest average salary:")
print(dept_avg.max())

print("\nOverall average salary:")
print(avg_salary)

print("\nEmployees earning above average:")
print(df[df['Salary'] > avg_salary][
    ['Name', 'Department', 'Salary']
])
```
