```python
import os
import csv
ip = []
http_status = []


path = "C:\\Users\\Muthu selvakumar\\Desktop"

filename = "mylog.log"

with open(os.path.join(path,filename)) as mydata:
     for each_line in mydata:
         mylist = each_line . split(" ")
         ip.append(mylist[3])
         http_status.append(mylist[8])
                
print(ip)
print()
print( http_status)

def frequency(mylist):
    mydict = {}
    for each in mylist:
        if each not in mydict:
           mydict[each] = 1
        else:
            mydict[each] +=1 
    return mydict
                  
print(frequency(ip))
print()
print(frequency(http_status))

def write_to_csv(mydict):
    with open(os.path.join(path,"muthu2.csv"),"w",newline ="")as mydata:
        writer = csv.writer(mydata)
        writer.writerow(["Sno","Ip","Frequency"])
        Sno = 1
        for each in mydict:
            writer.writerow([Sno,each,mydict[each]])
            Sno +=1
            
x = frequency(http_status)
print(x)
write_to_csv(x)


import pandas as pd
import matplotlib.pyplot as plt
import os


desktop = os.path.join(os.path.expanduser("~"), "Desktop")

df = pd.read_csv(os.path.join(desktop, "muthu1.csv")
plt.figure(figsize=(12, 6))

plt.bar(df["Ip"], df["Frequency"])

plt.title("IP Address Frequency")
plt.xlabel("IP Address")
plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(os.path.join(desktop, "IP_Frequency.png"))

plt.show()





import pandas as pd
import matplotlib.pyplot as plt
import os


desktop = os.path.join(os.path.expanduser("~"), "Desktop")

df = pd.read_csv(os.path.join(desktop, "muthu2.csv"))

plt.figure(figsize=(10, 5))

plt.bar(df["Ip"], df["Frequency"])

plt.title("HTTP Status Code Frequency")
plt.xlabel("HTTP Status Code")
plt.ylabel("Frequency")

# Show values on bars
for i, value in enumerate(df["Frequency"]):
    plt.text(df["Ip"].iloc[i], value, str(value), ha="center", va="bottom")

plt.tight_layout()

plt.savefig(os.path.join(desktop, "muthu2_graph.png"))

plt.show()

            
            
```
