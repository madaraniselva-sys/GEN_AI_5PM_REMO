```python
#For Loop
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
row=int(input("Enter no.of.rows :"))
for i in range(row):
    for j in range(row-i,0,-1):
        print(" ",end='')
    for k in range(i+1):
        print(Symbol,end='')
    print()
```

```python
Symbol = input("Give any Symbol: (!/@/#/$/%/^/&): ")
row = int(input("Enter no. of rows: "))
a = 1
while a <= row:
    b = 1
    while b <= row+1-a:
        print(" ", end='')
        b += 1
    c=1
    while c <= a:
        print(Symbol, end='')
        c += 1
    print()
    a += 1
```
