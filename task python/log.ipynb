## **get unic ip ID. from the log the request**
## **get how many 200,300,400 error code&status**
## **write the result to a csv file**

```python
import os
import csv
ip = []
http_status = []


path = "C:\\Users\\Muthu selvakumar\\Desktop"

filename = "mylog.log"

with open(os.path.join(path,filename)) as mydata:
     for each_line in mydata:
         mylist = each_line . split(" ")
         ip.append(mylist[3])
         http_status.append(mylist[8])
                
print(ip)
print()
print( http_status)
                
def frequency(mylist):
    mydict = {}
    for each in mylist:
        if each not in mydict:
           mydict[each] = 1
        else:
             mydict[each] = +1 
    return mydict
                  
print(frequency(ip))
print()
print(frequency(http_status)) 
 
def write_to_csv(mydict):
    with open(os.path.join(path, "new_csv.csv"), "w", newline= "") as mydata:
        writer = csv.writer(mydata)
        writer.writerow(["Sno", "IP", "Frequency"])
        Sno = 1
        for each in mydict:
            writer.writerow([Sno,each, mydict[each]])
            Sno += 1    
                
x = frequency(ip)
print(x)
write_to_csv(x)
                                              
```
