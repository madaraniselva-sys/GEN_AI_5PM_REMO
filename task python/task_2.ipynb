*WHILE LOOP PATTERN*

**pattern**

```python
#1-pattern
x=int(input("Enter no.of.rows :"))
y=int(input("Enter no.of.colums :"))
a=1
while a<=x:
    b=1
    while b<=y:
        print(b,end='')
        b+=1
    print()
    a+=1
```

```python
#2-pattern
row=int(input("Enter no.of.rows :"))
coloum=int(input("Enter no.of.colums :"))
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
a=1
while a<=row:
    b=1
    while b<=coloum:
        print(Symbol,end='')
        b+=1
    print()
    a+=1
    
```

```python
#3-pattern
row = int(input("Enter no. of rows: "))
a = 1
while a <= row:
    b = 1
    while b <= a:
        print(b, end='')
        b += 1
    print()
    a += 1
```

```python
#4-pattern
Symbol = input("Give any Symbol: (!/@/#/$/%/^/&): ")
row = int(input("Enter no. of rows: "))
a = 1
while a <= row:
    b = 1
    while b <= a:
        print(Symbol, end='')
        b += 1
    print()
    a += 1
```

**FOR LOOP PATTERN**

```python
#1-pattern
row=int(input("Enter no.of.rows :"))
coloum=int(input("Enter no.of.colums :"))
for i in range(row):
    for j in range(1,coloum+1):
        print(j,end='')
    print()
```

```python
#2-pattern
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
row=int(input("Enter no.of.rows :"))
coloum=int(input("Enter no.of.colums :"))
for i in range(row):
    for j in range(1,coloum+1):
        print(Symbol,end='')
    print()
```

```python
#3-pattern
row=int(input("Enter no.of.rows :"))
for i in range(row):
    for j in range(1,i+2):
        print(j,end='')
    print()
```

```python
#4-pattern
Symbol=input("Give any Symbol:(!/@/#/$/%/^/&) :")
row=int(input("Enter no.of.rows :"))
for i in range(row):
    for j in range(0,i+1):
        print(Symbol,end='')
    print()
```
