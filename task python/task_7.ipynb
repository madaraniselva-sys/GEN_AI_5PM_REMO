### **Python Assessment Questions**

**1.Write a Python program to find the length of a list using len().**

```python
numbers = [56789, 12345, 67890, 23456]
print("Length of the list:", len(numbers))
```

**2.Write a Python program to find the maximum and minimum value in a list using max() and min().**


```python
numbers = [10, 25, 5, 40, 15]

print("Maximum value:", max(numbers))
print("Minimum value:", min(numbers))
```

**3.Use sum() to calculate the total of all numbers in a list.**

```python
numbers = [10, 20, 30, 40, 50]
print("Total:", sum(numbers))
```

**4.Create a list of numbers and sort them in ascending and descending order using sort().**


```python
numbers = [40, 10, 30, 20, 50]

numbers.sort()
print("Ascending order:", numbers)

numbers.sort(reverse=True)
print("Descending order:", numbers)
```

**5.Write a Python program to create a tuple and print it.**

```python
my_tuple = (10, 20, 30, 40, 50)
print(my_tuple)
```

**6.Write a program to access the first, middle, and last elements of a tuple using indexing.**


```python
my_tuple = (10, 20, 30, 40, 50)

print("First element:", my_tuple[0])
print("Middle element:", my_tuple[len(my_tuple) // 2])
print("Last element:", my_tuple[-1])
```

**7.Write a program to slice a tuple (first 3 elements, last 3 elements).**

```python
my_tuple = (10, 44, 66, 88, 27, 45)

print("First 3 elements:", my_tuple[:3])
print("Last 3 elements:", my_tuple[-3:])
```

**8.Write a program to remove an element from a set using discard().**

```python
my_set = {10, 20, 30, 40, 50}

my_set.discard(30)
print(my_set)
```

**9.Write a program to pop a random element from a set using pop().**

```python
my_set = {10, 20, 30, 40, 50}

removed_element = my_set.pop()

print("Removed element:", removed_element)
print("Updated set:", my_set)
```

**10.Write a program to update the value of an existing key in a dictionary.**

```python
student = {"name": "muthu selva kumar", "age": 28, "city": "naraikinar"}

student["age"] = 25

print(student)
```

**11.Write a program to remove the last inserted key-value pair using popitem().**


```python
student = {"name": "muthu", "age": 60, "city": "Chennai"}

removed_item = student.popitem()

print("Removed item:", removed_item)
print("Updated dictionary:", student)
```

**12.Write a program to remove all key-value pairs using clear().**

```python
student = {"name": "muthu", "age": 28, "city": "NRK"}

student.clear()

print(student)
```

**13.Write a program to get all keys using keys().**

```python
student = {"name": "muthu", "age": 28, "city": "Chennai"}

print(student.keys())
```

**14.Write a program to get all key-value pairs using items().**

```python
student = {"name": "ranii", "age": 28, "city": "Chennai"}

print(student.items())
```

**15.Write a program to check if a key exists in a dictionary using in.**

```python
student = {"name": "Rani", "age": 54, "city": "Chennai"}

if "age" in student:
    print("Key exists")
else:
    print("Key does not exist")
```
